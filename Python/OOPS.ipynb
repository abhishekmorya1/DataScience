```python
class Student:

    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def talk(self):
        print("Hello", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)


name = input("Enter your name: ")
age = int(input("Enter your age: "))
marks = int(input("Enter your marks: "))

s = Student(name, age, marks)

s.talk()
```

```python

class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def volume(self):
        print("Volume:", 3.14 * self.radius ** 2 * self.height)

    def total_surface_area(self):
        print("Total Surface Area:", 2 * 3.14 * self.radius * (self.radius + self.height))

    def curved_surface_area(self):
        print("Curved Surface Area:", 2 * 3.14 * self.radius * self.height)


c = Cylinder(4, 5)

c.volume()
c.total_surface_area()
c.curved_surface_area()
```
