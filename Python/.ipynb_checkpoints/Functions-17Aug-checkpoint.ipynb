```python
# counting function 

def count1(s, str1):
    count = 0 
    for i in s:
        if i == str1:
            count = count + 1

    return count
```

```python
print(count1("abhishekaa" , "a"))
```

```python
# replace function 

def replace(s, old, new):
    result = ""

    for i in s:
        if i == old:
            result = result + new
        else:
            result = result + i

    return result
```

```python
replace("abhishek" , "a" , "u")
```

```python
# factorial of a given number 

def fact1(num):
    for i in range(num-1, 1 , -2):
        result = num*i 
    return result
```

```python
fact1(5)
```

```python
def check_number():
    x = int(input("Enter the number greater than 50: "))

    if x > 100:
        print("Wrong number")
    else:
        print("Wrong number, again type")
        check_number()


check_number()
```
