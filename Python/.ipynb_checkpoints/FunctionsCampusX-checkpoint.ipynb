# Functions in Python


```python
# A function is a reusable block of code that performs a specific task.

# Instead of writing the same code again and again, we put it inside a function and call it whenever needed.
```

```python
def isEven(num):
    if num % 2 == 0:
        return 'even'
    else:
        return 'odd'

isEven(66)
```

```python
# use this above function in a loop 

for i in range(1,11):
    print(i , "->", isEven(i))
```

```python
# More good function 

def isEven1(num):
    if type(num) == int:
        if num % 2 == 0:
          return 'even'
        else:
          return 'odd'
    else:
        print("Please Enter Integer Only!")
```

```python
isEven1('hell0')
```
