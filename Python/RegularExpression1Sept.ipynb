```python
# on match object we can call the following method 

# start()- > return the start index of the match 
# end() -> return end +1 index of the match 
# group() -> return the matched string 
```

```python
# finditer 

# returns an interator object which yields match object for every match 
# matcher -> pattern.finditer("abaadaba") 
```

```python
s = "abcadababc"

x = re.finditer("ab", s)

for i in x:
    print(i.group(), "--match is available at start index:" , i.start())
```

```python
s = "abcadababc"

x = re.finditer("ab", s)

for i in x:
    print(i.group(), "--match is available at end index:" , i.end())
```

```python
# character classes 

#[abc] ->  a or b or c 
# [^abc] -> except a and b and c 


#[a-z] -> any lower case alphabet symbol 
#[A-Z] -> any upper case alphabet symbol 

#[a-zA-Z] -> any alphabet symbol 

#[0-9]
#[a-za-z0-9]
#[^a-zA-Z0-9]
```

```python
import re 
matcher = re.finditer("[abc]", "abhi558ac")

for i in matcher:
    print(i.group(), "-match is available at end index" , i.end())
```

```python
# predefined character classes 

# \s -> space character 
# \S -> except space character 

#|
```

```python
s = "abcadab 679%$abc"

x = re.finditer("\d", s)

for i in x:
    print(i.group(), "--match is available at start index:" , i.end())
```

```python
s = "abhis&^%$#hek"

x = re.sub("[^a-z]" , "", s)
x
```
