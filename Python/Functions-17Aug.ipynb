```python
# counting function 

def count1(s, str1):
    count = 0 
    for i in s:
        if i == str1:
            count = count + 1

    return count
```

```python
print(count1("abhishekaa" , "a"))
```

```python
# replace function 

def replace(s, old, new):
    result = ""

    for i in s:
        if i == old:
            result = result + new
        else:
            result = result + i

    return result
```

```python
replace("abhishek" , "a" , "u")
```

```python
# table
# multiple number
# factorial 
# sum of n natural number
#fibonacci series 
```

```python
def table(num, i=1):
    if i > 10:
        return
    
    print(num, "x", i, "=", num * i)
    table(num, i + 1)


table(5)
```

```python
def factorial(n):
    if n == 0:
        return 1
    
    return n * factorial(n - 1)


print(factorial(5))
```

```python
def sum_n(n):
    if n == 0:
        return 0
    
    return n + sum_n(n - 1)


print(sum_n(5))
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    
    return fibonacci(n - 1) + fibonacci(n - 2)


n = 7

for i in range(n):
    print(fibonacci(i), end=" ")
```

```python
def table(num, i=1):
    if i > 10:
        return

    print(num, "x", i, "=", num * i)
    table(num, i + 1)


def multiple_tables(start, end):
    if start > end:
        return

    table(start)
    print()   # Har table ke baad space

    multiple_tables(start + 1, end)


multiple_tables(5, 10)
```

```python
keypad = [
    "abc",
    "def",
    "ghi",
    "jkl",
    "mno",
    "pqr",
    "stu",
    "vwx",
    "yz"
]

word = input("Enter word: ")

count = 0

for ch in word:
    for group in keypad:
        if ch in group:
            count = count + group.index(ch) + 1

print("Total key presses:", count)
```
