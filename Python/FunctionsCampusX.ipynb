# Functions in Python


```python
# A function is a reusable block of code that performs a specific task.

# Instead of writing the same code again and again, we put it inside a function and call it whenever needed.
```

```python
def isEven(num):
    if num % 2 == 0:
        return 'even'
    else:
        return 'odd'

isEven(66)
```

```python
# use this above function in a loop 

for i in range(1,11):
    print(i , "->", isEven(i))
```

```python
# More good function 

def isEven1(num):
    if type(num) == int:
        if num % 2 == 0:
          return 'even'
        else:
          return 'odd'
    else:
        print("Please Enter Integer Only!")


  # Why?

# • type(num) == int → Checks whether the input is an integer.
# • num % 2 == 0 → Checks whether the number is divisible by 2.
# • return 'even' → Returns 'even' if the number is even.
# • return 'odd' → Returns 'odd' if the number is odd.
# • else → Handles non-integer inputs.
```

```python
isEven1('hell0')
```

# Default Argument

```python
# Default Parameter 

def power(a=1 , b=1):
    return a**b


# power(2,3)
# power(9)
# power()


# Why?

# • Default parameters are parameters that have a default value.
# • If no argument is provided, the default value is used.
# • Here, a=1 and b=1 are default parameters.
# • If an argument is provided, it replaces the default value.
```

```python
power(2,3)
```

# Positional Argument

```python
# Positional Arguments

def student(name, age, course):
    print(name, age, course)

student("Abhishek", 25, "Python")


# Why?

# • Positional arguments are arguments passed to a function based on their position/order.
# • The first argument is assigned to the first parameter.
# • The second argument is assigned to the second parameter.
# • The order of arguments matters.
# • Changing the order can change the result or meaning.
# • Here, 10 is assigned to a and 20 is assigned to b.
```

# Keyword Argument

```python
# Keyword Arguments

def student(name, age):
    return name, age


student(age=25, name="Abhishek")


# Why?

# • Keyword arguments are arguments passed to a function using parameter names.
# • Arguments are assigned based on parameter names, not their position.
# • The order of keyword arguments does not matter.
# • Here, age=25 is assigned to age and name="Abhishek" is assigned to name.
# • Keyword arguments make function calls more readable and clear.
```

# *args and **kwargs

```python
# *args
# allows us to pass a variable number of non - keyword argunments to a function 


def multiply(*args):
    product = 1

    for i in args:
        product = product * i

    return product

# multiply(2, 3, 4)


# Why?

# • *args is used to pass a variable number of positional arguments to a function.
# • All positional arguments passed to the function are stored in a tuple called args.
# • We can pass any number of positional arguments using *args.
# • Here, each value in args is multiplied using a for loop.
# • *args allows a function to accept multiple positional arguments without defining them individually.
# • The name args is not mandatory, but * must be used.
```

```python
multiply(1,2,3,4,5)
```

```python
# **kwargs 

def display(**kwargs):

    for (key,value) in kwargs.items():
        print(key , '->', value)

# Why?

# • **kwargs is used to pass a variable number of keyword arguments to a function.
# • All keyword arguments passed to the function are stored in a dictionary called kwargs.
# • Each keyword is stored as a key and its value is stored as the corresponding value.
# • Here, key and value are accessed using the items() method.
# • **kwargs allows a function to accept multiple keyword arguments without defining them individually.
# • The name kwargs is not mandatory, but ** must be used.
```

```python
display(india =  'delhi', srilanka = 'colombo', nepa = 'kathmandu')
```

```python
# Points to remember while using *args and **kwargs

# • The order of arguments matters:
#   normal arguments -> *args -> **kwargs
# • The words "args" and "kwargs" are only conventions.
# • You can use any valid name of your choice, but * and ** must be used.
```

```python
# How Functions Are Executed in Memory?

def add(a, b):
    return a + b


x = add(10, 20)
x

# Why?

# • When a function is defined, Python creates a function object in memory.
# • The function code is stored in memory, but the function is not executed yet.
# • When the function is called, a new stack frame is created in memory for that function call.
# • The arguments are assigned to the parameters inside this stack frame.
# • The function body is then executed line by line.
# • Local variables created inside the function are stored in its stack frame.
# • The return value is sent back to the calling code.
# • After the function finishes, its stack frame is removed from memory.
# • The function object itself remains in memory as long as it is still referenced.
```

```python
# How Functions Are Executed in Memory? (Without return)

def greet(name):
    message = "Hello " + name
    print(message)


greet("Abhishek")


# Why?

# • When a function is defined, Python creates a function object in memory.
# • The function is not executed until it is called.
# • When the function is called, a new stack frame is created for that function call.
# • The argument "Abhishek" is assigned to the parameter name.
# • The local variable message is created inside the function's stack frame.
# • The function body is executed line by line.
# • print() displays the output but does not return a value to the caller.
# • If a function has no return statement, Python automatically returns None.
# • After the function finishes, its stack frame is removed from memory.
```
