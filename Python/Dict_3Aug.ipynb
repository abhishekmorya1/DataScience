```python
x1 = ["a", "b", "c"]
x2 = [10, 20, 30]

d = {}

for i in range(len(x1)):
    d[x1[i]] = x2[i]

print(d)
```

```python
table = {}

for i in range(2, 11):
    table[i] = [i * j for j in range(1, 11)]

for key, value in table.items():
    print(f"{key} : {value}")
```

```python
s = ["arun" , "dev", "ankit"]
d={}

for i in s:
    d[i] = i[::-1]

d
    
```

```python
s = ["ar683un", "de@#$v", "ank34"]

d = {}

for i in s:
    name = ""
    for ch in i:
        if ch.isalpha():
            name += ch
    d[i] = name

print(d)
```

```python
# user input dictionary 

d = {}

n = int(input("Enter number of items: "))

for i in range(n):
    key = input("Enter key: ")
    value = input("Enter value: ")
    d[key] = value

print(d)
```

```python
d = {"name":["nikita", "tine","riya"] , "marks":80}
d
```

```python
d = {"name": ["nikita", "tine", "riya"], "marks": 80}

d["name"] = [name.replace("i", "u") for name in d["name"]]

print(d)
```

```python
d = {"name": ["nikita", "tine", "riya"], "marks": 80}

for i in range(len(d["name"])):
    d["name"][i] = d["name"][i].replace("i", "u")

print(d)
```
