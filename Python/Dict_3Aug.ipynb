```python
d ={}

x = ["arun", "dev", "python", "ankit", "nita"]

for i in x:
    d[i] = len(i)

d
    
```

```python
s = "abchshdbbaskaaajkskdjl"

d = {}

for i in s:
    d[i] = s.count(i)

d
        
```

```python
x1 = ["a", "b", "c"]
x2 = [10, 20, 30]

d = {}

for i in range(len(x1)):
    d[x1[i]] = x2[i]

print(d)
```
