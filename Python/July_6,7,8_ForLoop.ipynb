```python
# prime number using for loop 

num = 4
count = 0

for i in range(1 , num+1):
    if num%i==0:
        count+=1

if(count == 2):
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
# car speed if -> <45 -> low speed
# if 45-60 -> average speed
# above 60 -> every inc in 8km/hr -> 1 demerit point
# 80 -> overspeed by 20 -> 2 demerit points
# 90 -> overspeed by 30 -> 3 demerit points

speed = int(input("Enter the speed of the car: "))

if speed < 45:
    print("Low speed")
elif speed >= 45 and speed <= 60:
    print("Average speed")
else:
    demerit_points = (speed - 60) // 8
    print("Overspeeding by", speed - 60, "km/hr")
    print("Demerit points:", demerit_points)
```

```python
# write a program to calculate the electricity bill(accept number of units from user) according to the following criteria:
# for first 100 units Rs. 0/unit
# next 100 units Rs. 5/unit
# above 200 units Rs. 10/unit

units = int(input("Enter the number of units consumed: "))
bill = 0

if units <= 100:
    bill = 0
elif units <= 200:
    bill = (units - 100) * 5
else:
    bill = 100 * 5 + (units - 200) * 10

print("The electricity bill is: Rs.", bill)
```

```python
# sum of n natural number using while loop 

sum = 0 

i = 1

while i<=10:
    sum = sum + i
    i = i + 1

print(sum)
```

```python
# Print all even numbers from 1 to 20.

i = 1
sum = 0

while i <= 20:
    if i % 2 == 0:
        sum = sum + i
    i = i + 1

print(sum)
```

```python
# Print all odd numbers from 1 to 20.

i = 1
sum = 0

while i <= 20:
    if i % 2 != 0:
        sum = sum + i
    i = i + 1

print(sum)
```

```python
# Print the multiplication table of a given number.

num = int(input("Enter the Number :"))

i = 1

while i<=10:
    print(num , "x" , i , "=" , num*i)
    i = i+1
    
    
```

```python
# Nested table 

start = int(input("Enter the Starting Number: "))
end = int(input("Enter the Ending Number: "))

while start <= end:
    i = 1

    while i <= 10:
        print(start, "x", i, "=", start * i)
        i = i + 1
    print()
    start = start + 1
```

```python
# Check whether a number is a palindrome. 

n = 55
original = n
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

if original == rev:
    print("Palindrome")
else:
    print("Not a Palindrome")

```
