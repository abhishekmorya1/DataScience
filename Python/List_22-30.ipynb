```python
#  min and max in list

x = [1, 3, 45, 3, 45, 99]

minimum = x[0]
maximum = x[0]

for i in x:
    if i < minimum:
        minimum = i
    if i > maximum:
        maximum = i

print("Minimum =", minimum)
print("Maximum =", maximum)
```

```python
# sum in nested list 

x = [21, 23, [4,5,20], 45]

total = 0

for i in x:
    if type(i) == list:
        for j in i:
            total += j
    else:
        total += i

print(total)
```
