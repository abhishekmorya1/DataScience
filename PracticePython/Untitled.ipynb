```python
# check palindrome string 

str1 = input("Enter the string")
s=""

for i in str1:
    s = i+s

if(str1 == s):
    print("Palindrome String")
else:
    print("Not a Palindrome String")
```

```python
# Count Vowels and Consonants 

str1 = input("Enter the string")
vowels = 0
constants = 0

for i in str1:
    if i == "a" or i=="e" or i=="i" or i=="o" or i=="u":
        vowels+=1
    else:
        constants+=1

print("vowels count is:", vowels)
print("Constants count is: ", constants)
        
        
```

```python
l = [[2,3,4,5],[3,6,5,7],[8,7,6,5]]
x1=[]
for i in l:
    x1.append(i[::-1])

x1
```

```python
l = [2,3,4,5,[3,6,5,7],[8,7,6,5]]

x1=[]

for i in l:
    if type(i) == list:
        for j in i:
            x1.append(j)
    else:
        x1.append(i)

x1
```
