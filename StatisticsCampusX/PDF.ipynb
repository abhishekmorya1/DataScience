```python
import pandas as pd
```

```python
import random
```

```python
# roll a single dice 

L=[]

for i in range(1000):
    a=random.randint(1,6)
    L.append(a)
```

```python
len(L)
```

```python
L[:5]
```

```python
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()
```

```python
s
```

```python
s.plot(kind="bar")
```

```python
# roll a two dice 

L=[]

for i in range(10000):
    a=random.randint(1,6)
    b=random.randint(1,6)
    L.append(a+b)
```

```python
L[:10]
```

```python
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()
```

```python
s
```

```python
s.plot(kind="bar")
```

```python
import numpy as np
```

```python
np.cumsum(s)
```

```python
np.cumsum(s).plot(kind="bar")
```

# How to PDF(Probability Density Function) in Data Science

```python
import seaborn as sns
```

```python
iris  = pd.read_csv("iris.csv")
```

```python
iris.head(5)
```

```python
sns.kdeplot(data = iris, x="SepalLengthCm" , hue = "Species")


# KDE Plot (Kernel Density Estimate)
# Used to visualize the distribution of numerical data
# It creates a smooth curve instead of bars like a histogram.

# sns.kdeplot(
#     data=iris,              # DataFrame
#     x="SepalLengthCm",      # Numerical column for X-axis
#     hue="Species"           # Separate curves for each category
# )

# Syntax:
# sns.kdeplot(data=DATAFRAME, x="NUMERICAL_COLUMN", hue="CATEGORY_COLUMN")

# data  → Specifies the DataFrame
# x     → Specifies the numerical column to plot
# hue   → Separates the plot based on categories
#         (e.g., Setosa, Versicolor, Virginica)

# Example:
# sns.kdeplot(data=iris, x="SepalLengthCm", hue="Species")

# Higher curve → More data is concentrated around that value/range
# Lower curve  → Less data is concentrated around that value/range
```

```python
sns.kdeplot(data = iris, x="SepalWidthCm" , hue = "Species")
```

# How to use CDF

```python
# KDE Plot
sns.kdeplot(data=iris, x="PetalWidthCm", hue="Species")

# ECDF Plot - CDF
sns.ecdfplot(data=iris, x="PetalWidthCm", hue="Species")
```
