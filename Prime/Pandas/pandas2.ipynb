# for loop in dataframe

```python
import pandas as pd 
import numpy as np
```

```python
# using Series

id = pd.Series([1,2,3,4,5])
name= pd.Series(["abhishek" , "karan", "krrish", "kunal", "himanshu"])
age = pd.Series([24,23,24,24,23])
marks = pd.Series([87,88,79,66,79])

df = pd.DataFrame({"id":id, "name":name, "age":age, "marks":marks})
print(df)
```

```python
for i in df:
    print(i)


# A DataFrame is iterable over its columns by default.
```

```python
for i in df["name"]:
    print(i)


# The loop picks one value at a time from the name column and stores it in i and print 
```

```python
df["name"][0]
```

```python
df["name"][4]
```

```python
for i in df.index:
    print(df["name"][i])

# This first loops over the indexes, then uses each index to retrieve the corresponding value from the "name" column.


# Use df.index when:

# You need the row index.
# You want to access multiple columns using the same row.


# Use df["name"] when:
    
# You only need the values from the name column.

# | If you want...                               | Use                                     |
# | -------------------------------------------- | --------------------------------------- |
# | Row indexes                                  | `df.index`                              |
# | Values from the `name` column                | `df["name"]`                            |
# | Values from multiple columns in the same row | `df.index` (or preferably `iterrows()`) |
# | Just one column's values                     | `df["name"]`                            |
```

```python
# single row single column
# single row multile column 


# multiple row single column ->  same value
# multiple row single column -> different value


# multiple row multiple column -> same value 
# multiple row multiple column -> different value 
```

```python
# using Series

id = pd.Series([1,2,3,4,5])
name= pd.Series(["abhishek" , "karan", "krrish", "kunal", "himanshu"])
age = pd.Series([24,23,24,24,23])
marks = pd.Series([87,88,79,66,79])

df1 = pd.DataFrame({"id":id, "name":name, "age":age, "marks":marks})
print(df1)
```

```python
# single row single column 

df1.loc[1, "name"] = "kareena"
df1
```

```python
#  multiple column single row

df1.loc[1, ["age", "marks"]] = [30, 100]
df1
```

```python
# multiple row single column 

df1.loc[[3,4], "name"] = "teena"
df1
```

```python
# multiple row single column -> different value

df1.loc[[3,4], "name"] = ["yogita", "urvashi"]
df1
```

```python
# multiple row multiple column 

df1.loc[[3,4], ["name","age"]] = ["yogita", 50]
df1
```

```python
# multiple row multiple column -> different value 

df1.loc[[3,4], ["name","age"]] = [["yogita", 50],["vidya",100]]
df1
```

```python
# how to delete rows , columns from a dataframe

# using Series

id = pd.Series([1,2,3,4,5])
name= pd.Series(["abhishek" , "karan", "krrish", "kunal", "himanshu"])
age = pd.Series([24,23,24,24,23])
marks = pd.Series([87,88,79,66,79])

df0 = pd.DataFrame({"id":id, "name":name, "age":age, "marks":marks})
print(df0)


# Why is row 0 still showing?

# Because drop() does not change the original DataFrame by default. It returns a new DataFrame.
```

```python
df0.drop(0)

# returns a new DataFrame with row 0 removed, but df0 itself remains unchanged.
```
