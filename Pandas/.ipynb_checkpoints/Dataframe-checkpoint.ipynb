# Dataframe Attribute and Methods

```python
# shape 

movies.shape

# .shape - Returns the dimensions of a DataFrame.

# Why?
# A DataFrame has two dimensions:
# • Rows
# • Columns
# .shape tells us the total number of rows and columns.

# Syntax:
# dataframe_name.shape
```

```python
# dtypes

movies.dtypes
ipl.dtypes


# .dtypes - Returns the data type of each column in a DataFrame.

# Why?
# A DataFrame can contain different types of data.
# (integers, floats, strings, booleans, dates, etc.).
# .dtypes helps us identify the data type of every column.


# Why different data types?
# Because each column stores different kinds of data.
# • Numeric columns → int64/float64
# • Text columns    → object
# • Boolean columns → bool
# • Date columns    → datetime64 (if converted)
```

```python
# index 

movies.index

# .index - Returns the row labels (index) of a DataFrame.

# Why?
# Every DataFrame needs an index to uniquely identify each row.
# If no custom index is provided, Pandas automatically creates a RangeIndex.
```

```python
# columns 

movies.columns

# .columns - Returns the column names of a DataFrame.

# Why?
# A DataFrame consists of rows and columns.
# .columns helps us view all column names.
```

```python
# values 

movies.values  

# .values - Returns all data of a DataFrame as a NumPy array.

# Why?
# Sometimes we need only the data, not the row index or column names.
# .values extracts only the DataFrame values.

```

```python
# head and tail 

movies.head()

# By default, head() displays the first 5 rows of the DataFrame.

# Purpose
# View the beginning of the dataset.
# Check whether the data has been loaded correctly.
# Verify column names and values.
```

```python
# To display the first 10 rows:

movies.head(10)
```

```python
movies.head(-5)

# It returns all rows except the last 5 rows.

# | Code            | Output                     |
# | --------------- | -------------------------- |
# | `data.head()`   | First 5 rows               |
# | `data.head(10)` | First 10 rows              |
# | `data.head(-5)` | All rows except the last 5 |
```

```python
movies.tail()

# By default, tail() displays the last 5 rows of the DataFrame.

# Purpose
# View the end of the dataset.
# Check the last records.
# Verify whether data is complete.
```

```python
ipl.tail(-5)

# All rows except the first 5
```

```python
# sample 

movies.sample()

# sample() returns random row(s) from the DataFrame.
# By default, it returns 1 random row.
# sample() selects rows randomly, so the output changes each time.
```

```python
movies.sample(4)

# It returns 4 random rows from the DataFrame.
```

```python
movies.sample(frac = .10)

# The frac parameter specifies the fraction (percentage) of rows to return.
# means 10% of the total rows.

# Every time you run the code, you may get different rows because they are selected randomly.
```

```python
movies.sample(frac = 1)

# Return all rows in random order

# | Method                   | Meaning                             |
# | ------------------------ | ----------------------------------- |
# | `data.sample(n=5)`       | Return exactly **5 random rows**    |
# | `data.sample(frac=0.10)` | Return **10% of the total rows**    |
# | `data.sample(frac=1)`    | Return **all rows** in random order |
```

```python
movies.info()

# df.info() is a Pandas method that gives a quick summary of your DataFrame. It tells you about the structure of the data, 
# not the statistical values.
```

```python
# df.describe() is a Pandas method that generates summary statistics for your DataFrame.

movies.describe()

# movies.describe(include='all')


# By default, it displays summary statistics for numerical columns only.
# Using include='all' displays statistics for both numerical and categorical (string) columns.
```

```python
ipl.describe()


# .describe() - Returns a statistical summary of all numeric columns in a DataFrame.

# Why?
# A DataFrame may contain many numeric columns.
# Instead of checking each column separately,
# .describe() summarizes them all at once.
```

```python
# isnull

movies.isnull().sum()

# .isnull() - Checks for missing (NaN) values in every cell of a DataFrame.

# Why?
# Real-world datasets often contain missing values.
# .isnull() helps identify which cells are missing.

# .sum() - Counts the number of missing values in each column.
```

```python
# duplicated

movies.duplicated().sum()
```

```python
student_dict1 = {
    'iq': [100, 90, 120, 80, 0, 0],
    'marks': [80, 70, 100, 50, 0, 0],
    'package': [10, 7, 14, 2, 0, 0]
}

df = pd.DataFrame(student_dict1)
df


# .duplicated() - Checks for duplicate rows in a DataFrame.

# Why?
# Datasets may contain repeated rows.
# .duplicated() identifies duplicate rows by comparing each row
# with the previous rows.

# .sum() - Counts the total number of duplicate rows.
```

```python
df.duplicated().sum()
```

```python
movies.rename(columns={"title_x":"abhishek"})

# .rename() - Renames column names in a DataFrame.

# Why?
# Sometimes column names are not meaningful or need to be changed.
# .rename() lets us rename one or more columns without changing the data.

# Syntax:
# dataframe_name.rename(columns={"old_name":"new_name"})
```

# Math Methods

```python
# sum -> axis argument

df.sum()

# .sum() - Returns the sum of numeric values in a DataFrame.

# Why?
# A DataFrame has rows and columns.
# The axis parameter tells Pandas whether to perform the operation
# column-wise or row-wise.

 # axis=0 (default) → Column-wise operation.
```

```python
# row-wise sum 

df.sum(axis=1)

# Why output is row-wise?
# Because axis=1 tells Pandas to move across the columns
# and calculate the sum for each row.
```

```python
df.mean()
```

```python
df.median()
```

```python
df.max()
```

```python
df.std()
```

```python
df.var()
```
