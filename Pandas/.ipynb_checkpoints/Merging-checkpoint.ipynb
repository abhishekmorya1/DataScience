```python
import pandas as pd 
import numpy as np
```

```python
courses =  pd.read_csv("courses.csv")
students =  pd.read_csv("students (1).csv")
nov =  pd.read_csv("reg-month1.csv")
dec =  pd.read_csv("reg-month2.csv")

matches = pd.read_csv("matches.csv")
delivery =  pd.read_csv("deliveries (1).csv")
```

```python
# pd.concat 

# pd.concat([nov,dec])
regs = pd.concat([nov,dec], ignore_index=True)
regs

# pd.concat() -> Combine DataFrames

# pd.concat() is a Pandas function used to combine two or more DataFrames along rows or columns.

# Why use it?
# → To combine separate DataFrames into one DataFrame.
# → Useful when data is split into multiple DataFrames, such as different months.

# Here:
# → pd.concat([nov, dec])
# → nov and dec are two DataFrames.
# → By default, they are combined row-wise (one below another).

# What happens?
# → All rows of nov come first.
# → Then all rows of dec are added below them.
# → Columns are matched by column names.

# Important:
# → Default axis=0 → row-wise combination.
# → axis=1 → column-wise combination.
# → If columns differ, Pandas creates missing values (NaN) where data is unavailable.
# → concat() returns a new DataFrame; it does not modify the original DataFrames.
```

```python
# Multiindex dataframe 

pd.concat([nov,dec], keys=['Nov', 'Dec'])

# MultiIndex DataFrame -> pd.concat() with keys

# pd.concat() with the keys parameter is used to combine DataFrames and create a MultiIndex.

# Why use it?
# → To identify which original DataFrame each row came from.
# → Useful when combining data from different months, categories, or sources.

# Here:
# → nov and dec are combined row-wise.
# → keys=['Nov', 'Dec'] creates an additional index level.
# → 'Nov' identifies rows coming from nov.
# → 'Dec' identifies rows coming from dec.

# What happens?
# → The DataFrame gets two index levels:
#    1. Outer index -> Nov / Dec
#    2. Inner index -> Original row index

# Important:
# → keys must match the number of DataFrames being concatenated.
# → The original index is preserved as the inner index.
# → This creates a MultiIndex DataFrame.
```

```python
multi = pd.concat([nov,dec], keys=['Nov', 'Dec'])
multi.loc['Dec']
```

```python
# Horizontally concat the data 
pd.concat([nov,dec], axis=1)
```
