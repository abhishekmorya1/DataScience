```python
#head and tail 

subs.head()

# .head() - Returns the first 5 rows of a Series or DataFrame.

# Why?
# Large datasets can have thousands of rows.
# .head() lets you quickly preview the data without displaying the entire dataset.
```

```python
vk.head(3)

# View first 10 rows:
# subs.head(10)

# Why head(10)?
# Passing 10 tells Pandas to display the first 10 rows instead of the default 5.

# Key Points:
# • Default: .head() = first 5 rows.
# • .head(n) = first n rows.
# • Works with both Series and DataFrame.
# • Very useful for inspecting large datasets.
```

```python
# sample 

movies.sample()

# .sample() - Returns a random row(s) from a Series or DataFrame.

# Why?
# Sometimes we want to inspect random data instead of only the first or last rows.
# .sample() helps us get a random sample from the dataset.
```

```python
movies.sample(5)


# Why sample(5)?
# Passing 5 tells Pandas to return 5 random rows.

# Key Points:
# • Default: .sample() = 1 random row.
# • .sample(n) = n random rows.
# • Each execution may return different rows.
# • Works with both Series and DataFrame.
```

```python
# value_counts

movies.value_counts()

# .value_counts() - Counts the frequency of each unique value in a Series.

# Why?
# Sometimes values are repeated in a dataset.
# .value_counts() tells us how many times each unique value appears.

# Why output is sorted?
# Because .value_counts() sorts the result in descending order by default
# (highest frequency to lowest).

# Why Movie A = 3?
# Because 'Movie A' appears 3 times in the Series.

# Key Points:
# • Counts occurrences of each unique value.
# • Sorted in descending order by default.
# • Duplicate values are counted together.
# • Return type: Series
```

```python
# sort_values ->inplace

# vk.sort_values()

vk.sort_values(ascending=False)


# .sort_values() - Sorts the values of a Series.

# Why?
# Data is often not in the required order.
# .sort_values() arranges the values in ascending or descending order.

# Why?
# By default, ascending=True, so values are sorted from lowest to highest.
```

```python
#  method changing

vk.sort_values(ascending=False).head(1).values[0]

# vk.sort_values(ascending=False).head(1).values[0]
# Returns the highest value in the Series.

# Why?
# We first sort the Series from highest to lowest.
# Then we take the first row.
# Finally, we extract only the value (not the index).
```

```python
# vk.sort_values(inplace=True)

# What is inplace?
# inplace=True modifies the original Series.
# inplace=False (default) returns a new sorted Series and leaves the original unchanged.
```

```python
# sort_index ->inplace -> movies 

movies.sort_index()

# .sort_index() - Sorts a Series by its index (labels).

# Why?
# Sometimes the index is not in the required order.
# .sort_index() arranges the index instead of the values.

# Why output is sorted?
# Because .sort_index() sorts the index labels in ascending order by default.
```

# Series Maths Methods

```python
# count 

vk.count()

# .count() - Counts the number of non-missing (non-NaN) values in a Series.

# Why?
# Datasets may contain missing values (NaN).
# .count() counts only the valid (non-NaN) values.
```

```python
# sum -> product 

subs.sum()  


# .sum() - Returns the sum of all values in a Series.

# Why?
# When we need the total of all numeric values,
# .sum() adds every value in the Series.
```

```python
# mean -> median -> mode -> std -> var

subs.mean()

print(vk.median())

print(movies.mode())

print(subs.std())

print(vk.var())
```

```python
# .mean() - Returns the average (arithmetic mean) of all values.

# Why?
# It helps find the central (average) value of the data.

# Formula:
# Mean = Sum of all values / Number of values


# .median() - Returns the middle value of the sorted data.

# Why?
# Mean can be affected by very large or very small values (outliers).
# Median gives the true center of the data.

# Syntax:
# series_name.median()



# .mode() - Returns the most frequently occurring value(s).

# Why?
# It helps identify the value(s) that appear most often.

# Syntax:
# series_name.mode()


# .std() - Returns the standard deviation of the data.

# Why?
# It measures how spread out the values are from the mean.
# A larger standard deviation means greater variation.

# Syntax:
# series_name.std()



# .var() - Returns the variance of the data.

# Why?
# Variance measures the spread of the data.
# It is the square of the standard deviation.

# Formula:
# Variance = (Standard Deviation)²

# Syntax:
# series_name.var()






```

```python
# min/max 

subs.min()

# .min() - Returns the smallest value in a Series.

# Why?
# When we need the lowest value in the dataset,
# .min() finds and returns it.


subs.max()

# .max() - Returns the largest value in a Series.

# Why?
# When we need the highest value in the dataset,
# .max() finds and returns it.
```

# Series Indexing

```python
# integer Indexing 

x = pd.Series([12,344,55,32,66,78])

x[0]


# Integer Indexing - Accesses a value using its integer index (position).

# Why?
# Every Series has an index.
# If no custom index is provided, Pandas automatically assigns
# integer indexes starting from 0 (0, 1, 2, ...).
```

```python
# Negative Indexing

# x[-1]

#  pandas Series doesn't support negative indexing
```

```python
# slicing   

vk[5:15]


# Slicing - Returns multiple values from a Series using a range of positions.

# Why?
# Instead of accessing one value at a time,
# slicing lets us retrieve multiple consecutive values.
```

```python
vk[-5:]


# | Code       | Returns                      | Why?                            |
# | ---------- | ---------------------------- | ------------------------------- |
# | `vk[5:15]` | Positions 5 to 14            | Stop index (15) is excluded     |
# | `vk[:5]`   | First 5 values               | Starts from the beginning       |
# | `vk[5:]`   | From position 5 to the end   | No stop value given             |
# | `vk[-5:]`  | Last 5 values                | Starts 5 positions from the end |
# | `vk[:-5]`  | All except the last 5 values | Stops before the last 5         |
```

```python
movies[-5:]


# Single indexing (vk[5]) → Accesses one element.
# Slicing (vk[5:15]) → Accesses multiple elements.
# In slicing, start is included and stop is excluded ([start, stop)).
```

```python
# fancy Indexing 

vk[[1,4,6,3]]

# Fancy Indexing - Selects multiple specific elements using a list of index labels.

# Why?
# Sometimes we don't need consecutive values (like slicing).
# We need only specific values from different indexes.
```

```python
#  indexing with labels -> fancy Indexing 

movies['2 States (2014 film)']


# Indexing with Labels - Accesses a value using its index label.

# Why?
# If a Series has custom index labels (e.g., movie names),
# we can access values directly using those labels instead of positions.

# Why output is 'Arjun Kapoor'?
# Because '2 States (2014 film)' is the index label,
# and its corresponding value is 'Arjun Kapoor'.
```

# Editing Series 

```python
# using Indexing 

dict_marks[1] = 100

dict_marks
```

```python
# what if an index does not exist 

dict_marks['jdjd'] = 52

dict_marks

# Adding a New Index - Assigns a value to a new index label.

# Why?
# If the specified index label does not exist,
# Pandas automatically creates a new index and assigns the value.


# Why is 'jdjd' added?
# Because the index label 'jdjd' did not exist,
# so Pandas created a new index and stored the value 52.


# | Code                       | Result          | Why?                 |
# | -------------------------- | --------------- | -------------------- |
# | `dict_marks['jdjd'] = 52`  | New index added | Label doesn't exist  |
# | `dict_marks['maths'] = 95` | Value updated   | Label already exists |
```

```python
# fancy indexing

run_Series[[0,3,4]] = [0,0,0]
run_Series


# Fancy Indexing Assignment - Replaces values at specific index positions.

# Why?
# Sometimes we need to update non-consecutive elements.
# Fancy indexing lets us select multiple specific indexes at once.

# Syntax:
# series_name[[index1, index2, index3]] = [value1, value2, value3]
```

```python
# using index label 

movies["Hum Tumhare Hain Sanam"] = "Abhishek Morya"
movies


# Label Assignment - Adds a new value or updates an existing value using an index label.

# Why?
# Every value in a Series is identified by an index label.
# Assigning a value to a label lets you add a new entry or update an existing one.

# Syntax:
# series_name["index_label"] = value
```

```python
# astype 

import sys 

sys.getsizeof(vk)


# sys.getsizeof() - Returns the memory size of an object in bytes.

# Why?
# Every object occupies memory.
# sys.getsizeof() tells us how much memory the object uses.
```

```python
sys.getsizeof(vk.astype('int16'))


# astype() - Converts the data type of a Series.

# Why?
# Smaller data types use less memory.
# Converting int64 to int16 reduces memory usage.

# Syntax:
# series_name.astype("new_dtype")


# Why output is smaller?
# Because the original Series uses int64 (8 bytes per value),
# while int16 uses only 2 bytes per value.
# Therefore, the converted Series occupies less memory.

# Key Points:
# • astype() changes the data type of a Series.
# • Smaller data types can reduce memory usage.
# • astype() returns a new Series; the original remains unchanged.
# • Return type: Series
```

```python
vk.isnull().sum()

# .isnull() - Checks whether each value is missing (NaN).

# Why?
# Real-world datasets often contain missing values.
# .isnull() helps identify them.

# Syntax:
series_name.isnull()
```

```python
# | Method            | Counts                   | NaN Included?      |
# | ----------------- | ------------------------ | ------------------ |
# | `.size`           | Total elements           | ✅ Yes              |
# | `.count()`        | Valid (non-NaN) values   | ❌ No               |
# | `.isnull()`       | Marks missing values     | Returns True/False |
# | `.isnull().sum()` | Number of missing values | ✅ Counts only NaN  |
```

```python
# dropna

temp1.dropna()


# .dropna() - Removes all missing (NaN) values from a Series.

# Why?
# Missing values can affect data analysis and calculations.
# .dropna() removes them and returns only valid values.


# Key Points:
# • Removes all NaN values.
# • Original indexes are preserved.
# • Returns a new Series (original is unchanged).
# • Return type: Series
```

```python
# fillna 

temp1.fillna(0)

# .fillna() - Replaces missing (NaN) values with a specified value.

# Why?
# Missing values can cause problems during analysis.
# .fillna() replaces them instead of removing them.

# Syntax:
# series_name.fillna(value)

# Example:
# temp1.fillna(0)
```
