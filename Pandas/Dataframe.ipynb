```python
# apply on dataframe 

points_df = pd.DataFrame(
    {
        '1st point':[(3,4),(-6,5),(0,0),(-10,1),(4,5)],
        '2nd point':[(-3,4),(0,0),(2,2),(10,10),(1,1)]
    }
)

points_df
```

```python
def euclidean(row):
    pt_A = row['1st point']
    pt_B = row['2nd point']

    return ((pt_A[0] - pt_B[0])**2 + (pt_A[1] - pt_B[1])**2)**0.5
```

```python
points_df['distance'] = points_df.apply(euclidean, axis=1)
points_df
```

```python
# isin(series)

import pandas as pd

students = pd.DataFrame({
    "name": ["Amit", "Rahul", "Priya", "Neha", "Vikas", "Anjali"],
    "branch": ["CSE", "ECE", "IT", "CSE", "ME", "IT"],
    "marks": [85, 72, 91, 65, 55, 88]
})

students
```

```python
students[students['branch'].isin(['CSE', 'IT'])]

# .isin() - Checks whether values belong to a given list.

# Why?
# It allows us to filter rows where a column contains
# any value from a specified list.
```

```python
students[students['marks'].isin([91])]

# .isin() - Filters rows by checking whether values
# are present in a given list.


# Key Points:
# • .isin() checks membership in a list.
# • Returns True/False for each row.
# • Used for filtering.
# • One value → similar to ==.
# • Multiple values → much cleaner than multiple OR conditions.
```

```python
# corr() 

import pandas as pd

df = pd.DataFrame({
    "Hours_Studied": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Sleep_Hours": [8, 7, 6, 5, 4]
})

df

# Key Points:
# • Measures linear relationship between numeric columns.
# • Range: -1 to +1.
# • +ve → Variables increase together.
# • -ve → One increases while the other decreases.
# • 0 → No linear relationship.
# • Returns a DataFrame.
```

```python
df.corr()

# .corr() - Calculates the correlation between numeric columns.

# Why?
# It tells us how strongly two numeric variables are related.
# The correlation value ranges from -1 to +1.
```

```python
# | Correlation | Meaning                        |
# | ----------: | ------------------------------ |
# |        `+1` | Perfect positive relationship  |
# |      `+0.8` | Strong positive relationship   |
# |      `+0.5` | Moderate positive relationship |
# |         `0` | No linear relationship         |
# |      `-0.5` | Moderate negative relationship |
# |      `-0.8` | Strong negative relationship   |
# |        `-1` | Perfect negative relationship  |
```

```python
# nlargest and nsmallest(series and dataframe)
```

```python
import pandas as pd

marks = pd.Series([45, 89, 72, 95, 61, 88, 76])

marks
```

```python
marks.nlargest(3)

# .nlargest() - Returns the largest N values from a Series.

# Why?
# Sometimes we need only the top values instead of sorting
# the entire Series.
```

```python
marks.nsmallest(3)

# .nsmallest() - Returns the smallest N values from a Series.

# Why?
# Sometimes we need the lowest values without sorting
# the entire Series.
```

```python
students = pd.DataFrame({
    "name": ["Amit", "Rahul", "Priya", "Neha", "Vikas"],
    "marks": [85, 72, 95, 65, 88],
    "age": [20, 21, 19, 22, 20]
})

students
```

```python
students.nlargest(3, "marks")

# .nlargest() - Returns the rows with the largest values
# from a specified DataFrame column.

# Why?
# In a DataFrame, we need to specify which column
# should be used to find the largest values.
```

```python
students.nsmallest(3, "marks")
```

```python
students.nlargest(
    3,
    ["marks", "age"]
)
# Pandas first sorts by marks, and if two students have the same marks,
# it uses age to break the tie.

# .nlargest() with multiple columns - Returns the top N rows
# based on multiple columns.


# Why?
# Step 1 → Highest marks are selected.
# Step 2 → If marks are equal, higher age gets priority.
# Step 3 → Top 3 rows are returned.
```

```python
# copy() in Pandas — Series + DataFrame

# copy() is used when you want to create a separate copy of a Pandas Series or DataFrame.

# The most important reason to understand copy() is to avoid accidentally changing the original data.
```

```python
import pandas as pd

marks = pd.Series([80, 90, 70, 60])

marks
```

```python
marks2 = marks.copy()
```

```python
marks2

# Changing marks2 does not affect marks.
```

```python
# copy() creates an independent copy of a Pandas Series or DataFrame, so modifications made to the copy do not affect the original object.
```
