```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv("imdb-top-1000 (1).csv")
movies.head(2)
```

```python
genres = movies.groupby('Genre')
```

```python
# find director with most popularity 

movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# first()/last() -> nth item
genres1 = movies.groupby('Genre')
genres1.first()
```

```python
genres1.last()
```

```python
# 7th movie of each genre 

genres1.nth(6)

# nth() is a Pandas GroupBy method used to retrieve the nth row from each group.

# Why use it?
# → To get a specific position/row from every group.
```

```python
# get_group -> vs filtering 

genres1.get_group("Fantasy")


# What:
# get_group() is a Pandas GroupBy method used to retrieve all rows belonging to a specific group.

# Why:
# It is used when you want to access the complete data of one particular group after using groupby().
```

```python
genres1.groups
```

```python
genres1.describe()
```

```python
genres1.sample(2, replace=True)
```

```python
genres1.nunique()
```

```python
# agg method 
# passing dict 

genres1.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum',
        'Metascore':'min'
    }
)


# agg() -> Aggregate data of each group

# agg() (aggregate) is a Pandas GroupBy method used to apply different aggregation functions to different columns of each group.

# Why use it?
# → To calculate multiple summary values for each group at once.
# → Different columns can use different aggregation functions.

# Passing a dictionary:
# → Dictionary keys = column names
# → Dictionary values = aggregation function to apply on that column
```

```python
# passing list 

genres1[['Runtime', 'IMDB_Rating', 'No_of_Votes', 'Gross', 'Metascore']].agg(
    ['min', 'max', 'mean']
)
```

```python
# merging both the syntax


genres1.agg(
    {
        'Runtime':['mean','sum'],
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum',
        'Metascore':'min'
    }
)
```

```python
# find the highest rated movie of each genre using for loop

# looping on groups
df = pd.DataFrame(columns=movies.columns)

for group, data in genres:
    df = pd.concat([
        df,
        data[data['IMDB_Rating'] == data['IMDB_Rating'].max()]
    ])

df
```

```python
# finding number of movies starting with A for each group 
  
def foo(group):
    return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(foo)
```

```python
# find ranking of each movie in the group according to IMDB score 

def rank_movie(group):
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending = False)
    return group
```

```python
genres1.apply(rank_movie)
```

```python
# find normalized IMDB rating group wise

def normal(group):
  group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group

genres.apply(normal)
```

```python
ipl = pd.read_csv("deliveries (2).csv")
ipl
```

```python
ipl.shape
```

```python
# find the top 10 batsman in terms of runs
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no of sixes
six = ipl[ipl['batsman_runs'] == 6]

six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
# find batsman with most number of 4's and 6's in last 5 overs
temp_df = ipl[ipl['over'] > 15]
temp_df = temp_df[(temp_df['batsman_runs'] == 4) | (temp_df['batsman_runs'] == 6)]
temp_df.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
# find V Kohli's record against all teams
temp_df = ipl[ipl['batsman'] == 'V Kohli']

temp_df.groupby('bowling_team')['batsman_runs'].sum().reset_index()
```
