```python
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle = 'solid', linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dotted', linewidth =2)

plt.title("Rohit Sharma VS Virat Kohle career comparison")
plt.xlabel('Season')
plt.ylabel('Runs Scored')


# Changing Line Width

## Definition
# The `linewidth` parameter of `plt.plot()` is used to control the **thickness of a plotted line**.

## Why?
# It helps make important lines more visible and improves the readability of graphs containing multiple lines.

## Syntax
# plt.plot(x, y, linewidth=value)
```

# Stacked Bar Chart 



```python
# # Stacked Bar Chart

# What:
# A stacked bar chart is a bar chart where multiple values are placed on top of each other within the same bar.

# Why:
# It is used to compare the total value of each category and also see how different components contribute to that total.

# How:
# Each bar represents a category, and different sections of the same bar represent different components.

# Result:
# The total height of each bar represents the combined value of all components.

# Important:
# Stacked bar charts are useful when you want to show both individual contributions and their overall total.
```

```python
plt.bar(batsmen["batsman"], batsmen["2017"], label = "2017")
plt.bar(batsmen["batsman"], batsmen["2016"], bottom=batsmen['2017'], label = "2016")
plt.bar(batsmen["batsman"], batsmen["2015"],bottom=batsmen['2016'] + batsmen["2017"], label = "2015")

plt.legend()
plt.show()

    # Stacked Bar Chart Using bottom


# What:
# The bottom parameter is used to place one bar on top of another and create a stacked bar chart.

# Why:
# It allows multiple datasets to be displayed as different sections of the same bar.

# How:
# The first bar is plotted normally. The next bar uses bottom to specify where it should start.

# bottom=batsmen['2015']:
# The 2016 values start from the top of the 2015 bars.

# Important:
# For proper stacking, the bottom of each new layer should be the cumulative total of all previous layers.

# In the given code:
# 2015 → First layer
# 2016 → Starts from 2015
# 2017 → Should start from 2015 + 2016, not only 2016

# So, bottom=batsmen['2016'] for 2017 will place 2017 from the top of 2016 values, rather than from the cumulative top of 2015 + 2016.
```

# Pie Chart

```python
# # Pie Chart

# univariate/Bivariate Analysis
# Categorical vs numerical 
# use case -> to find contribution on a standard scale 

# What:
# A pie chart is used to represent parts of a whole using circular slices.

# Why:
# It is used to show the proportion or percentage contribution of different categories to a total.

# How:
# The complete circle represents 100% of the data, and each slice represents one category's share of the total.

# Syntax:
# plt.pie(data)

# Result:
# Larger values create larger slices, while smaller values create smaller slices.

# Important:
# The values should represent parts of a meaningful whole.
```

```python
# simple data 

marks = [25,45,99,55,43]
subjects = ["english" , "hindi", "maths", "science", "sst"]

plt.pie(marks, labels = subjects)

plt.show()
```

```python
# colors 

plt.pie(gayle["batsman_runs"], labels = gayle["batsman"], autopct='%0.1f%%', colors=['blue','green', 'yellow', "pink", "cyan","brown"] , explode=[0.1, 0, 0, 0, 0, 0], shadow=True )
plt.show()


# Pie Chart Colors, Explode and Shadow

# colors:
# Used to specify the color of each slice in the pie chart.

# explode:
# Used to separate or highlight specific slices by moving them slightly away from the center.

# explode=[0.1, 0, 0, 0, 0, 0]:
# Moves the first slice outward by 0.1. The remaining slices stay in their original position.

# shadow=True:
# Adds a shadow effect behind the pie chart to give it a 3D-like appearance.

# Important:
# The number of colors and explode values should match the number of slices in the pie chart.

# colors → Controls slice colors
# explode → Highlights specific slices
# shadow → Adds shadow effect
```
